# LAP solver results: plots

Regenerates the runtime, speedup, energy, FLOPs, convergence and LOBPCG figures from the result files in `data/`.

Paths are set in the next cell. Defaults are relative to this notebook; each can be overridden with an environment variable:

| Variable | Env var | Default | Contents |
|---|---|---|---|
| `DATA_DIR` | `LAP_DATA_DIR` | `data` | result CSVs (`LOBPCG.csv`, `MI300x_LOBPCG_results.csv`, ...) |
| `LOBPCG_DIR` | `LAP_LOBPCG_DIR` | `data/MI300x_LOBPCG` | raw MI300x LOBPCG outputs: `<matrix>/output_results/<preconditioner>.txt` |
| `OUTPUT_DIR` | `LAP_OUTPUT_DIR` | `plots` | where figures are written |
| `MATRIX_DIR` | `LAP_MATRIX_DIR` | `matrices` | optional `.mtx` files, used only by the sparsity-pattern cells at the end |


```python
import os

DATA_DIR = os.path.abspath(os.environ.get('LAP_DATA_DIR', 'data'))
LOBPCG_DIR = os.path.abspath(os.environ.get('LAP_LOBPCG_DIR', os.path.join(DATA_DIR, 'MI300x_LOBPCG')))
OUTPUT_DIR = os.path.abspath(os.environ.get('LAP_OUTPUT_DIR', 'plots'))
MATRIX_DIR = os.path.abspath(os.environ.get('LAP_MATRIX_DIR', 'matrices'))

os.makedirs(OUTPUT_DIR, exist_ok=True)
print('DATA_DIR  ', DATA_DIR)
print('LOBPCG_DIR', LOBPCG_DIR)
print('OUTPUT_DIR', OUTPUT_DIR)
```

```python

# ============================================================
# MI300x LOBPCG: Extract execution times from result files,
# save CSV, and plot normalized runtime w.r.t. GS_std
# ============================================================
import os, re, glob

base_dir = LOBPCG_DIR

# Preconditioner file names -> display names (consistent with other plots)
prec_file_map = {
    'GS_std':       'Gauss-Seidel',
    'GS_it':        'Two Stage Gauss-Seidel (GS_it)',
    'GS_it2':       'Jacobi + Two Stage Gauss-Seidel (GS_it2)',
    'line_jacobi':  'Jacobi',
    'it_jacobi':    'iterative Jacobi',
    # 'ichol':      'Incomplete Cholesky',   # provision
    # 'none':       'None',                  # provision
}

def parse_lobpcg_result(filepath):
    """Extract time, iterations, converged count from a LOBPCG output file."""
    time_s, iters, converged = None, None, None
    try:
        with open(filepath, 'r') as f:
            text = f.read()
        m_time = re.search(r'Time \(seconds\)\s*:\s*([\d.]+)', text)
        m_iter = re.search(r'Iterations\s*:\s*(\d+)', text)
        m_conv = re.search(r'Converged\s*:\s*(\d+)\s*/\s*(\d+)', text)
        if m_time:
            time_s = float(m_time.group(1))
        if m_iter:
            iters = int(m_iter.group(1))
        if m_conv:
            converged = f"{m_conv.group(1)}/{m_conv.group(2)}"
    except Exception:
        pass
    return time_s, iters, converged

# Collect results
results = []
for matrix_dir in sorted(os.listdir(base_dir)):
    out_dir = os.path.join(base_dir, matrix_dir, 'output_results')
    if not os.path.isdir(out_dir):
        continue
    for prec_file, prec_name in prec_file_map.items():
        fpath = os.path.join(out_dir, prec_file + '.txt')
        if os.path.isfile(fpath):
            time_s, iters, converged = parse_lobpcg_result(fpath)
            results.append({
                'Matrix': matrix_dir,
                'Preconditioner': prec_name,
                'Prec_key': prec_file,
                'Time_s': time_s,
                'Iterations': iters,
                'Converged': converged,
            })

df_mi300x = pd.DataFrame(results)

# Special case: road_usa GS_std is truncated/missing; user says ~26h = 93600s
mask_road_gs = (df_mi300x['Matrix'] == 'road_usa') & (df_mi300x['Prec_key'] == 'GS_std')
if mask_road_gs.any() and pd.isna(df_mi300x.loc[mask_road_gs, 'Time_s'].values[0]):
    df_mi300x.loc[mask_road_gs, 'Time_s'] = 93600.0
    df_mi300x.loc[mask_road_gs, 'Iterations'] = -1
    df_mi300x.loc[mask_road_gs, 'Converged'] = '~26h (est.)'

# Save raw CSV
csv_path = os.path.join(DATA_DIR, 'MI300x_LOBPCG_results.csv')
df_mi300x.to_csv(csv_path, index=False)
print(f"Saved {csv_path}")
print(df_mi300x.to_string(index=False))

# --- Normalized runtime w.r.t. GS_std ---
gs_times = df_mi300x[df_mi300x['Prec_key'] == 'GS_std'].set_index('Matrix')['Time_s']

norm_records = []
for _, row in df_mi300x.iterrows():
    gs_t = gs_times.get(row['Matrix'])
    if pd.notna(row['Time_s']) and pd.notna(gs_t) and gs_t > 0:
        norm_records.append({
            'Matrix': row['Matrix'],
            'Preconditioner': row['Preconditioner'],
            'Normalized Runtime (w.r.t. GS)': row['Time_s'] / gs_t,
        })

df_norm = pd.DataFrame(norm_records)

prec_order = [
    'Gauss-Seidel',
    'Two Stage Gauss-Seidel (GS_it)',
    'Jacobi + Two Stage Gauss-Seidel (GS_it2)',
    'Jacobi',
    'iterative Jacobi',
]
palette_mi300x = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

has_data = set(zip(df_norm['Matrix'], df_norm['Preconditioner']))
LOW_VALUE_THRESHOLD = 0.15

sns.set_theme(style="darkgrid", font_scale=1.9)
g = sns.catplot(x="Preconditioner", y="Normalized Runtime (w.r.t. GS)",
                hue="Preconditioner", col="Matrix",
                data=df_norm, kind="bar",
                height=4, aspect=0.7, legend_out='False',
                palette=palette_mi300x, sharey=False,
                order=prec_order, hue_order=prec_order)

pre_1 = mpatches.Patch(color='#1f77b4', label='Gauss-Seidel')
pre_2 = mpatches.Patch(color='#ff7f0e', label='Two Stage Gauss-Seidel (GS_it)')
pre_3 = mpatches.Patch(color='#2ca02c', label='Jacobi + Two Stage Gauss-Seidel (GS_it2)')
pre_4 = mpatches.Patch(color='#d62728', label='Jacobi')
pre_5 = mpatches.Patch(color='#9467bd', label='iterative Jacobi')

g.set(xlabel=None)
g.set(xticklabels=[])
g.set_titles(template='{col_name}')
g.set_ylabels(fontsize=21)

for ax in g.axes.flat:
    ax.axhline(1.0, color='gray', linestyle='--', linewidth=1.2, alpha=0.8)
    matrix_name = ax.get_title()
    ylim_bot, ylim_top = ax.get_ylim()

    for i, prec_name in enumerate(prec_order):
        if (matrix_name, prec_name) not in has_data:
            ax.text(i, ylim_top * 0.03, 'no data',
                    ha='center', va='bottom', fontsize=8,
                    color='dimgray', fontstyle='italic', rotation=90)

    data_bars = [p for p in ax.patches if p.get_width() > 0.05 and np.isfinite(p.get_height()) and p.get_height() > 0]
    for patch in data_bars:
        height = patch.get_height()
        if height < LOW_VALUE_THRESHOLD:
            ax.text(patch.get_x() + patch.get_width() / 2.0,
                    height + (ylim_top - ylim_bot) * 0.01,
                    f'{height:.3f}',
                    ha='center', va='bottom', fontsize=9, color='black')

plt.tight_layout(pad=0.05)
plt.savefig(os.path.join(OUTPUT_DIR, "MI300x_lobpcg_runtime_normalized.pdf"), bbox_inches="tight")
plt.show()

# --- Separate legend ---
fig_leg, ax_leg = plt.subplots(figsize=(14, 0.5))
ax_leg.axis('off')
ax_leg.legend(handles=[pre_1, pre_2, pre_3, pre_4, pre_5], loc='center',
              ncol=5, fontsize=11, frameon=False)
plt.tight_layout(pad=0.05)
plt.savefig(os.path.join(OUTPUT_DIR, "MI300x_lobpcg_legend.pdf"), bbox_inches="tight")
plt.show()
```

```python

# ============================================================
# MI300x LOBPCG: Convergence curves per eigenvalue (thermal2)
# 4 subplots — one per non-trivial eigenvalue (2–5)
# ============================================================
import re, os

base_dir = LOBPCG_DIR
conv_matrix = 'thermal2'  # change to plot a different matrix

# Preconditioners to plot (file stem -> short legend label)
prec_conv_map = {
    'GS_std':       'GS',
    'GS_it':        'GS_it',
    'GS_it2':       'GS_it2',
    'line_jacobi':  'Jacobi',
    'it_jacobi':    'it Jacobi',
    'ichol':        'Ichol',
    # 'none':       'None',   # provision
}

# Matching colors
prec_colors = {
    'GS':        '#1f77b4',
    'GS_it':     '#ff7f0e',
    'GS_it2':    '#2ca02c',
    'Jacobi':    '#d62728',
    'it Jacobi': '#9467bd',
    'Ichol':     '#8c564b',
    # 'None':    '#7f7f7f',
}

def parse_convergence(filepath, num_eig=5):
    """Parse per-eigenvalue residual history from a LOBPCG output file.

    Returns dict: {eig_idx: [(iter, residual), ...]} for eig_idx 1..num_eig
    """
    history = {i: [] for i in range(1, num_eig + 1)}
    with open(filepath, 'r') as f:
        current_iter = None
        for line in f:
            # Match iteration header: "it    3  max residual = ..."
            m_it = re.match(r'^it\s+(\d+)\s+max residual', line)
            if m_it:
                current_iter = int(m_it.group(1))
                continue
            # Match eigenvalue line: "  eigenvalue 2: 1.23e-02  residual: 4.56e-03"
            m_ev = re.match(r'^\s+eigenvalue\s+(\d+):\s+[\d.e+\-]+\s+residual:\s+([\d.e+\-]+)', line)
            if m_ev and current_iter is not None:
                eig_idx = int(m_ev.group(1))
                residual = float(m_ev.group(2))
                history[eig_idx].append((current_iter, residual))
    return history

# Parse all preconditioners for the chosen matrix
conv_data = {}  # {prec_label: {eig_idx: [(iter, res), ...]}}
out_dir = os.path.join(base_dir, conv_matrix, 'output_results')
for prec_file, prec_label in prec_conv_map.items():
    fpath = os.path.join(out_dir, prec_file + '.txt')
    if os.path.isfile(fpath):
        conv_data[prec_label] = parse_convergence(fpath)

# Determine which labels are actually present
active_labels = [l for l in prec_conv_map.values() if l in conv_data]
active_palette = [prec_colors[l] for l in active_labels]

# --- Same seaborn style as existing convergence plots ---
sns.set_theme(style="dark", font_scale=1.7)

fig, axes = plt.subplots(1, 4, figsize=(20, 4.5), sharey=True)

for ax_idx, eig_idx in enumerate(range(2, 6)):
    ax = axes[ax_idx]

    # Build wide-format DataFrame with Iteration # as index
    # (like the CSVs: one column per preconditioner, values = log10 residual)
    df_eig = None
    for prec_label in active_labels:
        pts = conv_data[prec_label].get(eig_idx, [])
        if not pts:
            continue
        iters, residuals = zip(*pts)
        df_p = pd.DataFrame({'Iteration #': iters, prec_label: [np.log10(r) for r in residuals]})
        df_p = df_p.set_index('Iteration #')
        if df_eig is None:
            df_eig = df_p
        else:
            df_eig = df_eig.join(df_p, how='outer')

    if df_eig is not None:
        df_eig = df_eig.sort_index()
        # sns.scatterplot with index as x-axis (matching existing convergence cells)
        sns.scatterplot(data=df_eig, palette=active_palette,
                        linewidth=0, alpha=0.5, ax=ax)

    ax.set_xlabel('LOBPCG Iteration #')
    if ax_idx == 0:
        ax.set_ylabel('Residual(log10)')
    else:
        ax.set_ylabel('')
    ax.set_title('')  # no title on top
    # Place eigenvalue label at middle of top border
    ax.text(0.5, 0.95, f'eigenvalue {eig_idx}', transform=ax.transAxes,
            fontsize=18, fontweight='bold', va='top', ha='center',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    # Remove per-subplot legends (we'll add one shared legend)
    if ax.get_legend() is not None:
        ax.get_legend().remove()

# Shared horizontal legend across the top
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', fontsize=16,
           ncol=len(labels), framealpha=0.9, bbox_to_anchor=(0.5, 1.08))

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, f"MI300x_lobpcg_convergence_{conv_matrix}.pdf"),
            bbox_inches="tight")
plt.show()
```

```python
from matplotlib.ticker import MaxNLocator

# ============================================================
# V100 CG: Speedup w.r.t. Gauss-Seidel
# Speedup = GS_time / method_time
# ============================================================

df_v100_cg_sp = pd.read_csv(os.path.join(DATA_DIR, 'V100_cg_runtime_updated_v2.csv'))

prec_order_v100cg = [
    'Gauss-Seidel',
    'Two Stage Gauss-Seidel (GS_it)',
    'Jacobi + Two Stage Gauss-Seidel (GS_it2)',
    'Jacobi',
    'Incomple Cholesky',
]
palette_v100cg = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# GS times per matrix
gs_times_v100cg = df_v100_cg_sp[df_v100_cg_sp['Preconditioner'] == 'Gauss-Seidel'].set_index('Matrix')['Execution Time (in seconds)']

speedup_records_v100cg = []
for _, row in df_v100_cg_sp.iterrows():
    gs_t = gs_times_v100cg.get(row['Matrix'])
    t = row['Execution Time (in seconds)']
    if pd.notna(t) and pd.notna(gs_t) and gs_t > 0 and t > 0:
        speedup_records_v100cg.append({
            'Matrix': row['Matrix'],
            'Preconditioner': row['Preconditioner'],
            'Speedup (w.r.t. GS)': gs_t / t,
        })

df_speedup_v100cg = pd.DataFrame(speedup_records_v100cg)
has_data_v100cg = set(zip(df_speedup_v100cg['Matrix'], df_speedup_v100cg['Preconditioner']))

sns.set_theme(style="darkgrid", font_scale=1.9)
g = sns.catplot(x="Preconditioner", y="Speedup (w.r.t. GS)",
                hue="Preconditioner", col="Matrix",
                data=df_speedup_v100cg, kind="bar",
                height=4, aspect=0.7, legend=False,
                palette=palette_v100cg, sharey=False,
                order=prec_order_v100cg, hue_order=prec_order_v100cg)

g.set(xlabel=None)
g.set(xticklabels=[])
g.set_titles(template='{col_name}')
g.set_ylabels(fontsize=21)

for ax in g.axes.flat:
    ax.axhline(1.0, color='gray', linestyle='--', linewidth=1.2, alpha=0.8)
    ax.yaxis.set_major_locator(MaxNLocator(integer=True))
    matrix_name = ax.get_title()
    ylim_bot, ylim_top = ax.get_ylim()

    for i, prec_name in enumerate(prec_order_v100cg):
        if (matrix_name, prec_name) not in has_data_v100cg:
            ax.text(i, ylim_top * 0.03, 'NC',
                    ha='center', va='bottom', fontsize=8,
                    color='dimgray', fontstyle='italic', rotation=90)

    data_bars = [p for p in ax.patches if p.get_width() > 0.05 and np.isfinite(p.get_height()) and p.get_height() > 0]
    for patch in data_bars:
        height = patch.get_height()
        ax.text(patch.get_x() + patch.get_width() / 2.0,
                height + (ylim_top - ylim_bot) * 0.01,
                f'{height:.1f}x',
                ha='center', va='bottom', fontsize=9, color='black')

plt.savefig(os.path.join(OUTPUT_DIR, "v100_cg_speedup.pdf"), bbox_inches="tight")
plt.show()
```
